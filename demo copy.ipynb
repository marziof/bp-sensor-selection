## Random Regular Graph - Sensors

```python
N=300
d=3
lam=0.3
delta=0.05
G=nx.random_regular_graph(d,N)
for (u,v) in G.edges():
    G.edges[u,v]['lambda'] = lam
if (N<500) : nx.draw(G)
```

```python
snn = np.array(simulate_one_detSIR(G, s_type = "delta", S = delta, mask = ["SI"], T_max=20))
```

```python
T = len(snn) - 1
print(T)
```

```python
from Analysis.gen import generate_contacts
contacts = generate_contacts(G,T,lam)
```

```python
from Analysis.gen import generate_sensors_obs
from Analysis.measures import ti_inferred, ti_random, SE, MSE, ti_star, OV, MOV, OV_rnd, MOV_rnd
import networkx as nx


rho_list = np.arange(0.0, 1.0, 0.1)
ov_rnd_list = []
mov_rnd_list = []
ov_rescaled_list = []
mov_rescaled_list = []

for rho in rho_list:
    n_obs = int(rho * N)
    obs = generate_sensors_obs(snn, o_type="n_obs", M=n_obs, T_max=T)
    fg_ = fg.FactorGraph(N,T,contacts,obs,delta=delta)
    fg_.update(maxit=1000)
    marg = fg_.marginals()

    M0 = np.array([1 - marg[:, 0],marg[:,0],np.zeros_like(marg[:,0])])
    ov0 = OV(snn[0], np.argmax(M0,axis=0))
    ov0_rnd = OV_rnd(snn[0], M0)
    ov_rescaled = (ov0 - ov0_rnd) / (1 - ov0_rnd)
    mov0 = MOV(M0)
    mov0_rnd = MOV_rnd(M0)
    mov_rescaled = (mov0 - mov0_rnd) / (1 - mov0_rnd)
    ov_rnd_list.append(ov0_rnd)
    mov_rnd_list.append(mov0_rnd)
    ov_rescaled_list.append(ov_rescaled)
    mov_rescaled_list.append(mov_rescaled)

plt.plot(rho_list, ov_rescaled_list, label="OV rescaled")
plt.plot(rho_list, mov_rescaled_list, label="MOV rescaled")
plt.xlabel("rho")
plt.ylabel("OV / MOV")
plt.legend()
plt.show()
```

```python
from Analysis.gen import generate_sensors_obs
from Analysis.measures import ti_inferred, ti_random, SE, MSE, ti_star, OV, MOV, OV_rnd, MOV_rnd
import networkx as nx

def closeness_selection(G, n_obs):
    if G is None:
        raise ValueError("Graph G must be provided for closeness selection.")
    centrality = nx.closeness_centrality(G)

    items = list(centrality.items())
    np.random.shuffle(items)  # FIX: break symmetry

    sorted_nodes = [n for n, _ in sorted(items, key=lambda x: x[1], reverse=True)]    
    #sorted_nodes = sorted(centrality, key=centrality.get, reverse=True)
    selected_sensors = set(sorted_nodes[:n_obs])
    return selected_sensors

rho_list = np.arange(0.0, 1.0, 0.1)
ov_rnd_list = []
mov_rnd_list = []
ov_rescaled_list = []
mov_rescaled_list = []

for rho in rho_list:
    n_obs = int(rho * N)
    selected_sensors = closeness_selection(G, n_obs)
    obs = generate_sensors_obs(snn, o_type="n_obs", M=n_obs, T_max=T, selected_nodes=selected_sensors)
    fg_ = fg.FactorGraph(N,T,contacts,obs,delta=delta)
    fg_.update(maxit=1000)
    marg = fg_.marginals()

    M0 = np.array([1 - marg[:, 0],marg[:,0],np.zeros_like(marg[:,0])])
    ov0 = OV(snn[0], np.argmax(M0,axis=0))
    ov0_rnd = OV_rnd(snn[0], M0)
    ov_rescaled = (ov0 - ov0_rnd) / (1 - ov0_rnd)
    mov0 = MOV(M0)
    mov0_rnd = MOV_rnd(M0)
    mov_rescaled = (mov0 - mov0_rnd) / (1 - mov0_rnd)
    ov_rnd_list.append(ov0_rnd)
    mov_rnd_list.append(mov0_rnd)
    ov_rescaled_list.append(ov_rescaled)
    mov_rescaled_list.append(mov_rescaled)

plt.plot(rho_list, ov_rescaled_list, label="OV rescaled")
plt.plot(rho_list, mov_rescaled_list, label="MOV rescaled")
plt.xlabel("rho")
plt.ylabel("OV / MOV")
plt.legend()
plt.show()
```
